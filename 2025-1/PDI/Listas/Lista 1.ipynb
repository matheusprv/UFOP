# Questões sobre Imagens

````python
def show(img_list, caption_list, Axis=False):
    n = len(img_list)
    f, ax = plt.subplots(1,n, figsize=(10+5*(n-1),10))
    for i in range(n):
        ax[i].imshow(img_list[i], cmap='gray')
        ax[i].set_title(caption_list[i])
        if not Axis:
            ax[i].set_axis_off()
````

<p>14) Trocar entre si os canais vermelho e azul. Ou seja, a imagem R,G,B vai ficar B,G,R. Faça de duas formas distintas</p>

````python
img = data.chelsea()

nimg = img.T
nimg = np.stack((nimg[2], nimg[1], nimg[0])).T
show([img, nimg], ['Original', 'Trocar'])

imgcopy = img.copy()
imgcopy = imgcopy[ :, : , [2 ,1, 0]]
show([img, imgcopy], ['Original', 'Trocar'])
````

<p> 15) Rotacionar uma imagem em 90 graus sem utilizar funções que realizem a rotação de forma explicita. Implemente de duas formas distintas sem usar laços de repetição</p>

````python
def rotate90(img):
    img = img.T # Transforma de altura, largura, profundidade para profundidade, largura, altura
    r, g, b = img[0].T, img[1].T, img[2].T # Extrai o r,g,b de cada profundidade e os rotaciona
    out = np.stack((r, g, b)).T # Empilha o RGB novamente e transpoe para trocar a profundidade e a largura novamente
    return out
````

````python
def rotate90_v2(img):
    return np.transpose(img, axes=(1, 0, 2))
````

````python
img = data.chelsea()
nimg = rotate90(img)
show([img, nimg], ['0 graus', '90 graus'])

nimg = rotate90_v2(img)
show([img, nimg], ['0 graus', '90 graus'])
````
